```python
import pandas as pd
```

```python
df = pd.read_csv('./viajes_surfistas.csv')
df
```

```python
# Ver los distintos países que hay en el dataset
df['homecountry'].unique()
```

```python
# Ver la cantidad de países
len(df['homecountry'].unique())
```

```python
agrupado = df.groupby('homecountry').agg({
    'homename' : 'count'
})

agrupado
```

```python
agrupadoOrdenado = agrupado.sort_values('homename', ascending=False)
agrupadoOrdenado
```

```python
top_4 = agrupadoOrdenado.head(4)
top_4
```

```python
df_otros = agrupadoOrdenado.drop(index=top_4.index.values)
df_otros
```

```python
df_otros['pais'] = 'Otros'
df_otros
```

```python
df_otros_agrupados = df_otros.groupby('pais').agg({
    'homename' : 'sum'
})
df_otros_agrupados
```

```python
df_otros_agrupados = df_otros_agrupados.reset_index()
df_otros_agrupados
```

```python
top_4 = top_4.reset_index()
top_4
```

```python
top_4.columns = ['pais', 'homename']
top_4
```

```python
df_final = pd.concat([top_4, df_otros_agrupados])
df_final
```

```python
df_final = df_final.set_index('pais')
df_final
```

```python
ax = df_final.plot.pie(
    y = 'homename',
    startangle = 90,
    autopct = '%1.1f%%',
    colormap = 'Pastel1'
)

ax.set_ylabel('')
ax.legend(
    title  = 'País de Surfistas',
    fontsize = 8,
    bbox_to_anchor = (1, 1)
)

ax.set_title('Porcentaje de Surfistas por País de Procedencia')
```
