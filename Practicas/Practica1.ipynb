```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./comparasion_algoritmos.csv')
df
```

```python
ax = df.plot(
    # Tipo de gráfica
    kind = 'bar',
    # Marcamos que el dataset sea el Eje X
    x = 'dataset',
    # Marcamos que columnas vamos a tener
    y = ['algoritmo1', 'algoritmo2'],
    # Delimitamos el tamaño de la gráfica
    figsize = (12, 5)
)

# Mostramos la cuadrícula
ax.grid(visible = True)

# Ponemos las etiquetas y los títulos
ax.set_xlabel('Dataset')
ax.set_ylabel('Rendimiento')
ax.set_title('Evaluación de Algoritmo 1 vs Algoritmo 2')

# Etiquetas de la leyenda
ax.legend(
    labels = ['Algoritmo 1', 'Algoritmo 2']
)

# La cuadrícula no se sobreponga a las barras
ax.set_axisbelow(True)
```
